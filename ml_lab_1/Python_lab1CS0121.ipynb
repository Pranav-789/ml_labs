```python
#use of a iterator
#iterator- returns numbers that starts with 25 and each and every time it is passed it gets squared.

class MyInit:    #for initialization
  def __iter__(self):
    self.a = 25
    return self

  def __next__(self):  # does the operation
    x = self.a
    self.a = self.a ** 2
    return x

obj1 = MyInit()
itr = iter(obj1)

print(next(itr))
print(next(itr))
print(next(itr))
```

```python
def total(nums: list[int | float])-> int | float:
    ans = 0
    
    for num in nums:
        ans = ans + num

    return ans

def squared_total(nums: list[int | float]) -> int | float:
    ans = 0

    for num in nums:
        ans = ans + (num*num)

    return ans

def mean(nums: list[int | float]) -> int | float:
    n = len(nums)

    return total(nums)/n

def mode(nums: list[int | float]) -> int | float:
    freq = dict()
    maxf = 0
    mode_value = None

    for num in nums:
        freq[num] = freq.get(num, 0) + 1
        if freq[num] > maxf:
            mode_value = num
            maxf = freq[num]

    return mode_value

def median(nums: list[int | float]) -> int | float:
    n = len(nums)
    nums = sorted(nums)
    if n % 2 == 0:
        return (nums[n//2] + nums[n//2 - 1])/2
    else:
        return nums[n//2]

def variance(nums: list[int | float]) -> int | float:
    return (squared_total(nums) - (mean(nums)**2)*len(nums))/(len(nums)-1)

def std(nums: list[int | float]) -> int | float:
    return variance(nums)**(0.5)
```

```python
with open('poem.txt','r') as file: 
   
    # reading each line     
    len_list = []
    for line in file: 
        print(len(line))
        len_list.append(len(line))

    print("Sum: ", total(len_list))
    print("Mean: ", mean(len_list))
    print("Medain: ", median(len_list))
    print("std: ", std(len_list))
        
```

# Working with CSV

Let's import a file with fuel economy data

```python
import csv
with open('mpg.csv') as csvfile:
    d = list(csv.DictReader(csvfile))
```

Find out how many items are there in the dictionary

```python
print(len(d))
```

Find the column names with keys

```python
for column in d[0].keys():
    print(column)
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```python
sum = 0
for i in d:
    sum = sum + int(i['cty'])
print("average cty fuel economy: ", sum / len(d))
```

Find the average hwy fuel economy across all cars

```python
sum = 0
for i in d:
    sum = sum + int(i['hwy'])
print("average hwy fuel economy: ", sum / len(d))
```

Use set to return the unique values for the number of cylinders the cars.

```python
st = set()
for i in d:
    st.add(i['cyl'])
print(st)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```python
MpgByCyl = []

for c in st: # iterate over all the cylinder levels
    summpg = 0
    typecount = 0
    for i in d: # iterate over all dictionaries
        if i['cyl'] == c:      
            summpg += int(i['cty'])
            typecount += 1

    avg = summpg/typecount
    MpgByCyl.append((c, avg))

MpgByCyl.sort(key=lambda x: x[0])
print(MpgByCyl)
```

Return the unique values for the class types in our dataset using set.

```python
vehicleclass = set(i['class'] for i in d)
print(vehicleclass)
```

Find average miles per gallon in the highway for each class of vehicles

```python
MileageByClass = []

for t in vehicleclass: # iterate over all the vehicle classes
    sumpg = 0
    classCount = 0
    for i in d: 
        if i['class'] == t:       
            sumpg += int(i['hwy'])          
            classCount += 1
    MileageByClass.append((t, sumpg/classCount)) 

MileageByClass.sort(key=lambda x: x[1])
print(MileageByClass)
        
```
